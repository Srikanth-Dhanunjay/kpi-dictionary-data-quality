```python
import pandas as pd
df=pd.read_csv("../datasets/retail-orders-raw.csv")
print("dataset shape",df.shape)
df.head()
```

```python
print(df.dtypes)
```

```python
print(df["quantity"].unique())
print(df["unit_price"].unique())
print(df["discount_pct"].unique())
print(df["payment_status"].unique())
print(df["category"].unique())
print(df["city"].unique())
print(df["customer_segment"].unique())
print(df["order_date"].unique())
print(df["order_id"].unique())
```

```python
#missing values in each Attribute
missing_values=df.isnull().sum()
print(missing_values)

#calculating completeness percentage
completeness=(df.notnull().sum()/len(df))*100
print(completeness)
```

```python
#checking duplicate orders
duplicate_orders=df[df['order_id'].duplicated(keep=False)]
print(duplicate_orders)
```

```python
#valdiity for numeric fields
invalid_quantity = df[
    ~df["quantity"].astype(str).str.fullmatch(r"\d+")
]
print(invalid_quantity[["order_id", "quantity"]])
```

```python
invalid_discount=df[
    (df['discount_pct']<0)|(df['discount_pct']>100)
]
print(invalid_discount)
```

```python
parsed_dates=pd.to_datetime(df['order_date'],errors="coerce")
invalid_dates=df[
    df["order_date"].notna() & parsed_dates.isna()
]
print(invalid_dates)
```

```python
#consistency check 
for column in ['customer_segment','city','category','payment_status']:
    print(df[column].value_counts(dropna=False),"\n")
```

```python
print(df['order_date'].dropna().tolist())
```

```python
#check freshness of data i.e by date how recent is data
print("old date ",parsed_dates.min())
print("latest date ",parsed_dates.max())
```

```python
# Create a data quality summary

quality_summary = pd.DataFrame({
    "Dimension": [
        "Completeness",
        "Uniqueness",
        "Validity",
        "Consistency",
        "Freshness"
    ],
    "Finding": [
        "order_date, city and discount_pct contain missing values",
        "RT-1004 appears twice",
        "Invalid quantity (-1, two), discount (105), and date (2026-13-10)",
        "Student/student, Paid/paid, and mixed date formats",
        "Data ranges from 2026-01-03 to 2026-01-18"
    ]
})

print(quality_summary)
quality_summary.to_csv("../outputs/data_profile_summary.csv",index=False)
```

```python
#quality check order_id
order_id_missing=df['order_id'].isnull().sum() 
order_id_duplicates=df['order_id'].duplicated().sum()
print("Missing order_id count is ",order_id_missing)
print("Duplicate order_id count is ",order_id_duplicates)
```

```python
#quality check of order date
missing_dates=df['order_date'].isnull().sum()
invalid_order_dates=(
    (df['order_date'].notna())&(parsed_dates.isna())
).sum()
out_of_range_dates=(
    parsed_dates.notna()&(
        (parsed_dates<pd.Timestamp("2025-01-01")) | (parsed_dates>pd.Timestamp.today())
    )
).sum()
print("Missing dates count is ",missing_dates)
print("Invalid dates count is ",invalid_order_dates)
print("out of range dates count is ",out_of_range_dates)
```

```python
#quality check for customer segment 
missing_segments=df['customer_segment'].isnull().sum()
valid_segments=["Student","Fresher","Professional"]
normalized_segments=df['customer_segment'].str.strip().str.title()
invalid_segements=(
    normalized_segments.notna() & ~normalized_segments.isin(valid_segments)
).sum()
print("Missing segment values count is ",missing_segments)
print("invalid segment values count is ",invalid_segements)
```

```python
#quality check for city 
missing_city=df['city'].isnull().sum()
empty_city=(
    (df['city'].notna()) & (df['city'].str.strip()=="")
).sum()
print("Missing city count is ",missing_city)
print("empty city count is ",empty_city)
```

```python
#quality check for category 
missing_category=df['category'].isnull().sum()
valid_category=["Learning Kit","Course Access","Mentor Session"]
invalid_category=(
    df["category"].notna() & ~df["category"].isin(valid_category)
).sum()
print("Missing category count is ",missing_category)
print("Invalid category count is ",invalid_category)
```

```python
# Quality check: quantity
quantity_numeric = pd.to_numeric(df["quantity"], errors="coerce")
invalid_quantity = (
    quantity_numeric.isna() | (quantity_numeric <= 0)
).sum()
print("Invalid quantity values count is", invalid_quantity)
```

```python
# Quality check: unit_price
missing_unit_price = df["unit_price"].isnull().sum()
invalid_unit_price = (
    df["unit_price"].notna() & (df["unit_price"] < 0)
).sum()
print("Missing unit_price values count is ", missing_unit_price)
print("Invalid unit_price values count is ", invalid_unit_price)
```

```python
# Quality check: discount_pct
missing_discount = df["discount_pct"].isnull().sum()
invalid_discount = (
    df["discount_pct"].notna() &
    (
        (df["discount_pct"] < 0) | (df["discount_pct"] > 100)
    )
).sum()
print("Missing discount_pct values count is ", missing_discount)
print("Invalid discount_pct values count is ", invalid_discount)
```

```python
#quality check: payment status 
missing_payment=df['payment_status'].isnull().sum()
valid_payment=["Paid","Pending","Failed","Refunded"]
normalized_payment=df['payment_status'].str.strip().str.title()
invalid_payment=(
    normalized_payment.notna() & ~normalized_payment.isin(valid_payment)
).sum()
print("Missing payment status values count is ",missing_payment)
print("Invalid payment status values count is ",invalid_payment)
```

```python
# Consolidated data quality check results
quality_checks = pd.DataFrame({
    "Column": [
        "order_id",
        "order_date",
        "customer_segment",
        "city",
        "category",
        "quantity",
        "unit_price",
        "discount_pct",
        "payment_status"
    ],
    "Missing_Count": [
        order_id_missing,
        missing_dates,
        missing_segments,
        missing_city,
        missing_category,
        df["quantity"].isnull().sum(),
        missing_unit_price,
        missing_discount,
        missing_payment
    ],
    "Invalid_Count": [
        0,
        invalid_order_dates + out_of_range_dates,
        invalid_segements,
        empty_city,
        invalid_category,
        invalid_quantity,
        invalid_unit_price,
        invalid_discount,
        invalid_payment
    ]
})
print(quality_checks)
```

```python
# Show the dates that failed parsing
print(df.loc[
    df["order_date"].notna() & parsed_dates.isna(),
    ["order_id", "order_date"]
])
```

```python
# Check whether dates follow the required ISO YYYY-MM-DD format
iso_format = df["order_date"].fillna("").str.match(
    r"^\d{4}-\d{2}-\d{2}$"
)
invalid_iso_format = (
    df["order_date"].notna() & ~iso_format
).sum()
print("Invalid ISO date format count:", invalid_iso_format)
```

```python
#add pass or fail tags 
quality_checks['Status']=(
    (quality_checks['Missing_Count']==0) & (quality_checks['Invalid_Count']==0)
).map({True:"PASS",False:"FAIL"})
print(quality_checks)
quality_checks.to_csv("../outputs/data_quality_checks.csv",index=False)
```
